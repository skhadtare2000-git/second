```python
def decor(funct):
    def inner():
        print("*"*25) #before
        funct() #original func
        print("*"*25) # after
    return inner
@decor
def show():
    print("welcome to al nexus")
```

```python
def db_decor(funct):
    def inner():
        print("connetion established")
        funct()
        print("connetion terminated")
    return inner

@db_decor
def copper():
    print("data transfer for copper")

@db_decor
def silver():
    print("data transfer for silver")

@db_decor
def gold():
    print ("data transfer for gold")
```

```python
#Using a for Loop
num = int(input("Enter a number: "))

fact = 1

for i in range(1, num + 1):
    fact = fact * i

print("Factorial =", fact)
```

```python
#Using a while Loop
num = int(input("Enter a number: "))

fact = 1
i = 1

while i <= num:
    fact = fact * i
    i += 1

print("Factorial =", fact)
```

```python
#Using Recursion (Frequently Asked)
def factorial(n):

    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

num = int(input("Enter a number: "))

print("Factorial =", factorial(num))
```

```python
def prime_no():

    for num in range(2, 101):

        is_prime = True

        for i in range(2, num):

            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            yield num


for no in prime_no():
    print(no)
```
